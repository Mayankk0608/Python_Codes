### BFS 

```python
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    order = []

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            visited.add(vertex)
            order.append(vertex)

            for neighbor in graph[vertex]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return order 
```

```python

graph = {
    'A' : ['B', 'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B'],
    'F' : ['C'],
}

BFS_Order = bfs(graph, 'A')
print(f"BFS Traversal order: {BFS_Order}")
```

### Water Jub Problem

```python
def is_valid_state(x, y, X, Y):
    return 0 <= x <= X and 0 <= y <= Y

def water_jug_bfs(X, Y, Z):
    start = (0, 0)
    visited = set()
    queue = deque()
    queue.append((start, []))

    while queue:
        (jug1, jug2), path = queue.popleft()

        if jug1 == Z or jug2 == Z:
            return path + [(jug1, jug2)]
        
        if (jug1, jug2) in visited:
            continue
        visited.add((jug1, jug2))

        next_states = []

        next_states.append((X, jug2))
        next_states.append((jug1, Y))
        next_states.append((0, jug2))
        next_states.append((jug1, 0))

        pour = min(jug1, Y - jug2)
        next_states.append((jug1 - pour, jug2 + pour))

        pour = min(jug2, X - jug1)
        next_states.append((jug1 + pour, jug2 - pour))


        for state in next_states:
            if is_valid_state(state[0], state[1], X, Y) and state not in visited:
                queue.append((state, path + [(jug1, jug2)]))

    return None

X, Y, Z = 4, 3, 2
solution = water_jug_bfs(X, Y, Z)

if solution:
    print(f"Steps to Reach {Z} liters:")
    for step in solution:
        print(step)

else:
    print("No solution exists.")
```

### Punctuation Using NLP 

```python
text = "Konnichiwa, kore wa python no shitsumon desu!! Otskaresamadesu.."

tokens = []
current_word = ""

for char in text:
    if char in string.punctuation:
        if current_word:
            tokens.append(current_word)
            current_word = ""
        tokens.append(char)
    
    elif char.isspace():
        if current_word:
            tokens.append(current_word)
            current_word = ""

    else:
        current_word += char

if current_word:
    tokens.append(current_word)

print(tokens)
```
